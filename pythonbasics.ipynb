```python
#5. Functions
def check_ports(ports):
  if ports == 22:
    return "SSH"
  elif ports == 80:
    return"HTTP"
  elif ports == 443:
    return"HTTPS"
  elif ports == 25:
    return"SMTP"
  elif ports == 3389:
    return"RDP"
  elif ports == 21:
    return"FTP"
  else:
    return "Unknown"
for p in ports:
  print(p, check_ports(p))

print(check_ports(25))
print(check_ports(100))
```

LOG FILE DETECTOR

```python
def find_bruteforce(filename, threshold):
  failed_by_ip = {}

  with open(filename, "r") as f:
    for line in f:
      if "Failed" in line:
        parts = line.split()
        ip = parts[-1]
        failed_by_ip[ip] = failed_by_ip.get(ip, 0) + 1

  for ip, count in failed_by_ip.items():
    if count >= threshold:
      print("ALERT:", ip, "has", count, "failed logins")

find_bruteforce("auth.log", 3)
find_bruteforce("auth.log", 2)
```
